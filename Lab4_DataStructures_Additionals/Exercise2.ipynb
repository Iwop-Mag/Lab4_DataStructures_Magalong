```python
# Student-specific inputs
LAST_NAME = "Magalong"
STUDENT_ID = "TUPM-26-1134"
SEED_NUM = int(STUDENT_ID[-1])          # last digit of the student ID -> 4
FAVORITE_ARTIST = "TAYLOR SWIFT"        # <-- CHANGE THIS to your own favorite artist

# Character positions of FAVORITE_ARTIST (A=1, B=2, ... Z=26; spaces ignored)
ARTIST_POS = [ord(ch.upper()) - 64 for ch in FAVORITE_ARTIST if ch.isalpha()]

print("LAST_NAME      :", LAST_NAME)
print("SEED_NUM       :", SEED_NUM)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("ARTIST_POS     :", ARTIST_POS)
```

```python
from collections import defaultdict, Counter

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def make_score(k):
    """Score in the range 60-100 from SEED_NUM and artist character positions."""
    return 60 + (SEED_NUM * ARTIST_POS[k % len(ARTIST_POS)] + 3 * k) % 41

def get_category(score):
    if score >= 90: return "Outstanding"
    if score >= 85: return "Very Satisfactory"
    if score >= 75: return "Satisfactory"
    return "Needs Improvement"

records = []
k = 0
for st in students:
    for sub in subjects:
        records.append({"student": st, "subject": sub, "score": make_score(k)})
        k += 1

# Incomplete record: one student record with a missing score
records[7]["score"] = None

# Repeated student-subject record (re-assessment with a different score)
orig = records[12]
records.append({"student": orig["student"], "subject": orig["subject"], "score": (orig["score"] + 6 if orig["score"] <= 94 else orig["score"] - 6)})

print(f"{'#':<3}{'STUDENT':<15}{'SUBJECT':<14}SCORE")
for n, r in enumerate(records, 1):
    print(f"{n:<3}{r['student']:<15}{r['subject']:<14}{r['score']}")
```

```python
organized = defaultdict(lambda: defaultdict(list))
for r in records:
    organized[r["student"]][r["subject"]].append(r["score"])

for st in sorted(organized):
    print(st)
    for sub in subjects:
        print(f"   {sub:<12}: {organized[st][sub]}")
```

```python
pair_counts = Counter((r["student"], r["subject"]) for r in records)
repeated = {pair: n for pair, n in pair_counts.items() if n > 1}
incomplete = [(r["student"], r["subject"]) for r in records if r["score"] is None]

print("Repeated student-subject records:")
for (st, sub), n in repeated.items():
    print(f"  {st} / {sub}: {n} entries -> scores {organized[st][sub]}")
print("\nIncomplete records (missing score):")
for st, sub in incomplete:
    print(f"  {st} / {sub}")
```

```python
official = {}                       # (student, subject) -> latest score
for r in records:
    if r["score"] is not None:
        official[(r["student"], r["subject"])] = r["score"]

categorized = [(st, sub, sc, get_category(sc)) for (st, sub), sc in official.items()]

print(f"{'STUDENT':<15}{'SUBJECT':<14}{'SCORE':>6}  CATEGORY")
for st, sub, sc, cat in categorized:
    print(f"{st:<15}{sub:<14}{sc:>6}  {cat}")

print("\nCategory distribution:", dict(Counter(c[3] for c in categorized)))
```

```python
def summarize(scores):
    return {"count": len(scores), "avg": round(sum(scores) / len(scores), 2),
            "high": max(scores), "low": min(scores)}

by_student, by_subject = defaultdict(list), defaultdict(list)
for (st, sub), sc in official.items():
    by_student[st].append(sc)
    by_subject[sub].append(sc)

student_summary = {st: summarize(v) for st, v in by_student.items()}
subject_summary = {sub: summarize(v) for sub, v in by_subject.items()}
for d in (student_summary, subject_summary):
    for key, s in d.items():
        s["category"] = get_category(s["avg"])

print("--- Students ---")
for st, s in student_summary.items(): print(f"{st:<15}{s}")
print("\n--- Subjects ---")
for sub, s in subject_summary.items(): print(f"{sub:<14}{s}")

# Students with incomplete data
missing_by_student = {st: [sub for sub in subjects if (st, sub) not in official] for st in students}
print("\nMissing subjects per student:", {k: v for k, v in missing_by_student.items() if v})
```

```python
student_ranking = sorted(student_summary.items(), key=lambda kv: kv[1]["avg"], reverse=True)
subject_ranking = sorted(subject_summary.items(), key=lambda kv: kv[1]["avg"], reverse=True)
top_scores = sorted(categorized, key=lambda t: t[2], reverse=True)[:5]

print("Student ranking (by average):")
for i, (st, s) in enumerate(student_ranking, 1): print(f"  {i}. {st:<15}{s['avg']}")
print("\nSubject ranking (by average):")
for i, (sub, s) in enumerate(subject_ranking, 1): print(f"  {i}. {sub:<14}{s['avg']}")
print("\nTop 5 individual scores:")
for st, sub, sc, cat in top_scores: print(f"  {sc}  {st} / {sub}  ({cat})")
```

```python
line = "=" * 66
print(line)
print("FINAL STUDENT PERFORMANCE REPORT".center(66))
print(f"Prepared by: {LAST_NAME} | ID: {STUDENT_ID} | Seed: {SEED_NUM}".center(66))
print(line)
print(f"Records generated: {len(records)}   Valid (official): {len(official)}   "
      f"Repeated: {len(repeated)}   Incomplete: {len(incomplete)}\n")

print("STUDENT RANKING")
print(f"{'RANK':<5}{'STUDENT':<15}{'AVG':>7}{'HIGH':>6}{'LOW':>6}  CATEGORY")
for i, (st, s) in enumerate(student_ranking, 1):
    print(f"{i:<5}{st:<15}{s['avg']:>7}{s['high']:>6}{s['low']:>6}  {s['category']}")

print("\nSUBJECT RANKING")
print(f"{'RANK':<5}{'SUBJECT':<14}{'AVG':>7}{'HIGH':>6}{'LOW':>6}  CATEGORY")
for i, (sub, s) in enumerate(subject_ranking, 1):
    print(f"{i:<5}{sub:<14}{s['avg']:>7}{s['high']:>6}{s['low']:>6}  {s['category']}")

print("\nFLAGGED RECORDS")
for (st, sub), n in repeated.items():
    print(f"  REPEATED  : {st} / {sub} ({n} entries, latest used = {official[(st, sub)]})")
for st, sub in incomplete:
    print(f"  INCOMPLETE: {st} / {sub} (score missing, excluded)")
print(line)
```
